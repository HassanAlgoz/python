لكي نكتب نصًا برمجيًا بلغة بايثون نحتاج إلى أمرين مختلفين: **محرر النص البرمجي** (Code Editor) لكتابة الملف وتعديله وحذفه بسهولة، و**مفسر بايثون** (Python Interpreter) الذي يحوّل هذا النص إلى أوامر يفهمها الحاسب وينفّذها. فمثلًا قد يكون لدينا مجلد مشروع وفيه ملف `main.py`؛ الامتداد `.py` يدل على أن الملف مكتوب بلغة بايثون. أما Visual Studio Code فهو المحرر، وما نثبّته من بايثون فهو المفسر وليس «اللغة» ذاتها.

إذا ثبّتنا بايثون بالطريقة التقليدية من موقع [python.org](https://www.python.org/) صارت كل المشاريع على الجهاز مرتبطة **بنفس المفسر ونفس المكتبات**. فالمفسر له إصدار معيّن (مثل 3.9 أو 3.12)، والمكتبات مثل NumPy وSciPy وPandas لها إصدارات أيضًا. قد يعمل المشروع الأول على بايثون 3.9 ومكتبة بإصدار معيّن، ثم نرقّي أو نخفّض الإصدار في مشروع ثانٍ، فإذا عدنا للمشروع الأول وجدنا تعارضًا: خواص موجودة في إصدار وغير موجودة في آخر. لذلك لن نثبّت بايثون بهذه الطريقة المشتركة.

بدل ذلك نستخدم برنامجًا اسمه **uv**، وهو **مدير مشاريع بايثونية**. عبره نحمّل مفسر بايثون والمكتبات الخاصة بكل مشروع على حدة، فتُحفظ داخل مجلد مخفي يبدأ بنقطة (مثل `.venv`) يحتوي الملفات التنفيذية والمكتبات لذلك المشروع فقط، دون أن تتأثر المشاريع الأخرى. وفي مشروع ثانٍ يمكننا عبر `uv` أيضًا اختيار إصدار مختلف من بايثون ومكتبات مختلفة حسب حاجته. لذلك أول ما نثبّته هو `uv`، ثم عن طريقه نثبّت بايثون وبقية الحزم.

ثم **بيئة التطوير** (Development Environment) وهي الأدوات البرمجية الأساسية والمساعدة في كتابة النص البرمجي وتنفيذه واختباره وتصحيحه وما إلى ذلك.

أما **محرر النص البرمجي** (Code Editor) فسنختار Visual Studio Code التابع لشركة مايكروسوفت، وهو مختلف عن Visual Studio.

- وهو مفتوح المصدر (أي: يمكن الاطلاع على نصه البرمجي). 
- ويستعمله 73.6% من 58,121 متجاوب في [استبيان StackOverFlow 2024](https://survey.stackoverflow.co/2024/technology#1-integrated-development-environment) وما يميزه أنه سهل بالنسبة للمبتدئ، وقوي بالنسبة للمتمكن.
- وتُعنى الشركة بدعمه وصيانته باستمرار وتطويره، كما يساهم فيه آخرون من خارج الشركة لكوْنه مفتوح المصدر.
- وتم تصميمه كي يقبل الزيادات واللواحق البرمجية ليختار كل مبرمج ما يناسبه من الأدوات حسب احتياجاته.

موقع المحرر: [https://code.visualstudio.com/](https://code.visualstudio.com/)

نحمّل Visual Studio Code من صفحته الرئيسية أو من قسم التنزيل حسب نظام التشغيل، ثم نتبع خطوات التثبيت. في موقع المحرر صفحات مساعدة مثل **Get Started** وقسم **Python** مع **Quick Start** يفيدان في البداية.

بعد التثبيت نفتح مجلد مشروع عبر **Explorer** (المستكشف) باختيار **Open Folder**. داخل المجلد ننشئ ملفات مثل `main.py`؛ تظهر الملفات المفتوحة كـ **تبويبات** (Tabs) يمكن إغلاقها وإعادة فتحها. يمكننا كتابة أوامر مثل `print`، لكن تشغيلها يحتاج مفسر بايثون.

نثبّت **uv** من صفحة التثبيت الرسمية: على macOS وLinux ننسخ أمر التثبيت ونلصقه في **شاشة الأوامر** (Terminal) داخل VS Code (من القائمة View → Terminal، أو باختصار لوحة المفاتيح). بعد انتهاء التثبيت نتحقق بـ `uv --version`. ثم نحوّل المجلد إلى مشروع بايثوني بأمر `uv init .` فيظهر مثلًا ملف `.python-version` الذي يحدد إصدار بايثون. بعدها `uv sync` ينشئ **بيئة افتراضية** في المجلد المخفي `.venv` وفيه المفسر والمكتبات الخاصة بهذا المشروع فقط.

للتأكد أننا نستخدم مفسر المشروع وليس بايثون النظام، نختار من شريط الحالة المفسر `.venv/bin/python` (وغالبًا يظهر كـ Recommended). يمكن التحقق في شاشة الأوامر بـ `which python` لرؤية المسار داخل `.venv`. تشغيل الملف يعني عمليًا تمرير مسار المفسر ثم مسار الملف؛ زر **Run Python File** يفعل ذلك تلقائيًا. ويمكن تشغيل أسطر محددة بـ **Run Selection**، أو بـ Shift+Enter، أو التفاعل في شاشة أوامر بايثون التفاعلية؛ وعند الحاجة نمسح شاشة الأوامر ونفتح واحدة جديدة.

طريقة أخرى مفضّلة عند التجريب: وضع فاصل خلية `# %%` في ملف `.py` ثم تشغيل الخلية (**Run Cell**). لذلك نحتاج حزمة `ipykernel` عبر `uv add ipykernel`. ويمكن بدل ذلك إنشاء دفتر Jupyter بامتداد `.ipynb` (Interactive Python Notebook)، واختيار نفس نواة `.venv`، وتشغيل كل خلية على حدة؛ سياق التنفيذ متصل بين الخلايا ما دامت الجلسة قائمة، وترتيب التشغيل مهم، و**Restart** يفرّغ الذاكرة.

للتعديل السريع على الإعدادات والسمات نستخدم **لوحة الأوامر** (Command Palette) بـ Ctrl+Shift+P (مثل اختيار Color Theme). ومن قسم **Extensions** نثبّت الإضافات المفيدة، وأهمها إضافة Python التي تربط المحرر بالمفسر. للتفاصيل راجع توثيق المحرر نفسه.

في أعلى كثير من الدروس والتمارين ستجد رابط **تحميل الدفتر**. عند الضغط عليه يفتح المتصفح صفحةً تعرض محتوى الملف كنص JSON (وليس دفترًا تفاعليًا جاهزًا)، بهذا الشكل تقريبًا:

![صفحة JSON لدفتر Jupyter بعد فتح رابط التحميل](../assets/save_notebook_json.png){width=100%}

لحفظه كدفتر Jupyter:

1. اضغط `Ctrl+S` (أو من قائمة المتصفح: **Save Page As…** / **حفظ الصفحة باسم…**).
2. احفظ الملف بامتداد `.ipynb` (مثل `01_instructions.ipynb`). إن اقترح المتصفح امتدادًا آخر مثل `.txt` فغيّره يدويًا إلى `.ipynb`.
3. افتح الملف المحفوظ في Visual Studio Code، واختر نواة المشروع (`.venv`) كما سبق في إعداد بيئة التطوير.

بهذا يعمل الدفتر محليًا لديك بنفس طريقة أي ملف `.ipynb` آخر.

### الفرق بين `uv add` و `uv pip install`

| الأمر | أين يُستخدم؟ | ماذا يفعل؟ |
| --- | --- | --- |
| `uv add <package>` | محلياً في شاشة الأوامر، داخل مشروع `uv` | يثبت الحزمة **ويضيفها** إلى `pyproject.toml` (ويدير القفل `uv.lock`) |
| `uv pip install <package>` | بيئة مؤقتة مثل Colab، أو عندما تريد سلوكاً شبيهاً بـ `pip` | يثبت الحزمة في البيئة الحالية **دون** تعديل ملفات المشروع |

### علامة `!` في Jupyter و Colab

في خلية الشيفرة داخل Jupyter أو Colab، تُفسَّر الخلية كشيفرة بايثون افتراضياً. لتشغيل أمر شاشة الأوامر (shell) وليس بايثون، ضع علامة التعجب `!` في بداية السطر:

```text
! uv pip install geopy
```

بدون `!` سيحاول Jupyter تنفيذ الأمر كشيفرة بايثون فيفشل.

### الطريقة 1: محلياً (شاشة الأوامر)

في مجلد المشروع، من شاشة الأوامر:

```sh
uv add geopy ipykernel jupyturtle matplotlib pandas
```
